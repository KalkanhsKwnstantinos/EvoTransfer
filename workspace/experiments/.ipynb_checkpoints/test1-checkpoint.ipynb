Necessary packages

```python
import tensorflow as tf
from keras.layers import Input,Dense,Flatten
from keras.models import Sequential
import matplotlib.pyplot as plt
from keras.datasets import mnist
import numpy as np
```

Transformations

```python
#loading the dataset
(train_X, train_y), (test_X, test_y) = mnist.load_data()
```

Output

```python
#printing the shapes of the vectors 
print('X_train: ' + str(train_X.shape))
print('X_test:  '  + str(test_X.shape))

#Displaying the n first digits in the datset
for i in range(9):
  plt.subplot(330 + 1 + i) #rows col pos
  plt.imshow(train_X[i], cmap=plt.get_cmap('gray'))
plt.show()
```

```python
model = Sequential(
    [   Input((28,28)), 
        Flatten(),
        Dense(16, activation="relu", name="layer1"),
        Dense(16, activation="relu", name="layer2"),
        Dense(10, activation="softmax"),
    ]
)
```

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(x=train_X, y=train_y, epochs=25, validation_split=0.2)
```

```python
model.evaluate(x=test_X,y=test_y)
```

```python
print(np.argmax(model.predict(np.array([test_X[7]]))))
plt.imshow(test_X[7])
```
